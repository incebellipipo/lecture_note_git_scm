# 0. Setting up

## What git is for

Git answers four questions every engineering project eventually asks:

- What did this file look like last Tuesday, before I broke it?
- Who changed this number, and why?
- Can I try something risky without losing what works?
- How do two people change the same code without overwriting each other?

It answers them by storing **snapshots** of your project. The history is a
chain of those snapshots, and almost everything git does is adding to that
chain or moving along it.

## 3. A place to practise

Every chapter starts with this cell. It deletes the chapter's practice
folder, creates it again empty, and goes into it — so you can always start
a chapter over. It is plain Python rather than shell commands, so it works
the same on Windows, macOS and Linux.

````python
import os, shutil, stat, sys
from pathlib import Path


def remove_readonly(func, path, _):
    os.chmod(path, stat.S_IWRITE)  # git makes its object files read-only
    func(path)


work = Path.home() / "git-course" / "00-setup"
if work.exists():
    shutil.rmtree(work, onexc=remove_readonly)
work.mkdir(parents=True)
%cd {work}
````

## 4. The code we will version

A box-shaped barge: its immersed volume, and the mass of water it pushes
aside. To run it we write `"{sys.executable}"` instead of `python`: that is
the Python running this notebook, whatever it is called on your system. In
a terminal you would just type `python hydrostatics.py`.

````python
%%writefile hydrostatics.py
"""Hydrostatics for a box-shaped barge."""

RHO = 1025.0  # kg/m^3, sea water


def volume(length, beam, draft):
    """Immersed volume of the hull, in m^3."""
    return length * beam * draft


def displacement(length, beam, draft):
    """Mass of the water the hull pushes aside, in kg."""
    return RHO * volume(length, beam, draft)


if __name__ == "__main__":
    print(displacement(80, 14, 4.5), "kg")
````

````python
!"{sys.executable}" hydrostatics.py
````

And a file that checks the code is still correct:

````python
%%writefile test_hydrostatics.py
from hydrostatics import displacement, volume

assert volume(10, 4, 2) == 80
assert displacement(10, 4, 2) == 82000  # 80 m^3 of sea water

print("all checks passed")
````

````python
!"{sys.executable}" test_hydrostatics.py
````
